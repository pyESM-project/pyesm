# Tutorial - Production planning model (with non-linearities) 
Complete workflow for definining and solving a simple production planning under 
resource constraints with non-linearities.

Package import

```python
import cvxlab

model_dir_name = 'model_nonlinear' # or 'model_nonlinear'
```

Generation of model directory

```python
cvxlab.create_model_dir(
    settings_file_type='yml', # or xlsx
)
```

Fill model setup file(s) ...

Model class instance generation

```python
model = cvxlab.Model(
    model_dir_name=model_dir_name,
    log_level='debug',
    model_settings_from='yml', # or 'xlsx'
    use_existing_data=False,
    multiple_input_files=False,
    input_data_files_type="xlsx", # or "csv"
    detailed_validation=True,
)
```

Fill sets data (model coordinates) ...

Initialization of data structures

```python
model.initialize_model_environment()
```

Fill exogeneous model data ...

Initialization of numerical problem(s)

```python
model.refresh_database_and_initialize_problem()
```

Solution of numerical problem(s)

```python
model.run_model(
    integrated_problems=False, # False for 'model_nonlinear'
    solver='IPOPT', # 'IPOPT' for 'model_nonlinear'
    solver_settings={'nlp': True}
)
```

Export endogenous model data

```python
model.load_results_to_database()
```

```python
import cvxlab

model = cvxlab.Model(
    model_dir_name='model_decomposition',
    log_level='debug',
    model_settings_from='xlsx', # or 'xlsx'
    use_existing_data=False,
    multiple_input_files=False,
    input_data_files_type="xlsx", # or "csv"
    detailed_validation=True,
)
model.initialize_model_environment()
model.refresh_database_and_initialize_problem()
model.run_model(
    integrated_problems=True,
    solver='GUROBI',
)
model.load_results_to_database()
```
